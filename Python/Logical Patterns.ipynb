<h1>Logical Patterns</h1>
<p>Using simple logics with nested for loop we can design square patterns that plots characters in rows and columns</p>

<h3>Pattern 1</h3>


```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            print('*', end=' ')
        print()

if __name__ =='__main__':
    pattern()
```

<h3>Pattern 2</h3>

```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            print(i, end=' ') # here we are printing value of i
        print()

if __name__ =='__main__':
    pattern()
```

<h3>Pattern 3</h3>

```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            print(j, end=' ') # Now we are printing value of i
        print()

if __name__ =='__main__':
    pattern()
```

<h3>Pattern 4</h3>
<p><b>Unit Matrix:</b> When row index is equals to column index there is <code>1</code> and rest places are <code>0</code></p>

```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            if i==j:
                print('1', end = ' ')
            else:
                print('0', end = ' ')
        print()

if __name__ =='__main__':
    pattern()
```

<h3>Pattern 5</h3>

```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            if (i+j)%2==0:
                print('1', end = ' ')
            else:
                print('0', end = ' ')
        print()

if __name__ =='__main__':
    pattern()
```

<h3>Pattern 6</h3>
<p><b>Border Matrix:</b> There is 1 in all 4 borders and the inset values are 0</p>

```python
def pattern()->None:
    for i in range(1, 6):
        for j in range(1, 6):
            if i==1 or j==1 or i==5 or j==5:
                print('1', end = ' ')
            else:
                print('0', end = ' ')
        print()

if __name__ =='__main__':
    pattern()
```
